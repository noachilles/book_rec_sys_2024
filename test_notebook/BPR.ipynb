## BPR(Bayesian Personalized Ranking)

```python
import sys; sys.path.insert(0, '..')

from test_util.data_loader import DataLoader
from test_util.metric_calculator import MetricCalculator
```

```python
data_loader = DataLoader(num_users=1000, num_test_items=5, data_path='../data/ml-10m/')
movielens = data_loader.load()
```

```python
from test_src.bpr import BPRRecommender
recommender = BPRRecommender()
recommend_result = recommender.recommend(movielens)
```

```python
metric_calculator = MetricCalculator()
metrics = metric_calculator.calc(
    movielens.test.rating.tolist(), recommend_result.rating.tolist(),
    movielens.test_user2items, recommend_result.user2items, k=10
)
print(metrics)
```

```python
for factors in [5, 10, 30]:
    recommend_result = recommender.recommend(movielens, factors=factors)
    metrics = metric_calculator.calc(
        movielens.test.rating.tolist(), recommend_result.rating.tolist(),
        movielens.test_user2items, recommend_result.user2items, k=10
    )
    print(metrics)
```
