```python
import sys; sys.path.insert(0, '..')

from test_util.data_loader import DataLoader
from test_util.metric_calculator import MetricCalculator
```

```python
data_loader = DataLoader(num_users=1000, num_test_items=5, data_path='../data/ml-10m')
movielens = data_loader.load()
```

```python
user_movie_matrix = movielens.train.pivot(index='user_id', columns='movie_id', values='rating')
user_movie_matrix
```

```python
user_movie_matrix.fillna(0)
```

```python
from sklearn.decomposition import NMF
import numpy as np

user_id2index = dict(zip(user_movie_matrix.index, range(len(user_movie_matrix.index))))
movie_id2index = dict(zip(user_movie_matrix.columns, range(len(user_movie_matrix.columns))))
matrix = user_movie_matrix.fillna(movielens.train.rating.mean()).to_numpy()

nmf = NMF(n_components=5)
nmf.fit(matrix)
P = nmf.fit_transform(matrix)
Q = nmf.components_

pred_matrix = np.dot(P, Q)

print(f"P: {P.shape}, Q: {Q.shape}, pred_matrix: {pred_matrix.shape}")
```
